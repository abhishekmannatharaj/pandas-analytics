# Exploratory Data Analysis

```python
# Preview the top 5 rows and load dataset
import pandas as pd
import numpy as n
df = pd.read_csv('../data/raw/Sales.csv')
df.head()
```

```python
df.shape
```

```python
#Structural Dimensions & Schema Types
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
df.info()
```

```python
#Check for Missing/Null Values
null_counts = df.isnull().sum()
print("Null values per column:")
print(null_counts[null_counts > 0])
```

```python
#Statistics
df.describe()
```

```python
#unique values 
print("Unique Countries:", df['Country'].unique())
print("\nOrders per Country:")
print(df['Country'].value_counts())
```

```python
#group by Country and State to get total Revenue and Profit
sales_by_region = df.groupby(['Country', 'State'])[['Revenue', 'Profit']].sum().reset_index()
sales_by_region.sort_values(by='Revenue', ascending=False).head(10)
```
